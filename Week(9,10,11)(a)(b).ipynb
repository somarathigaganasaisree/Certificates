```python
#Finding quartiels
QQ1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Display the identified outliers.
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
outliers = df[
    (df["Age"] < lower_limit) |
    (df["Age"] > upper_limit)
]
print("\nDetected Outliers:")
print(outliers)
```

```python
# Original dataset
df = pd.read_csv("/hospital_patients_real_world.csv")

# Copy for dropping missing values
df_drop = df.copy()

# Copy for filling missing values
df_fill = df.copy()
```

```python
#Drop missing values:
df_drop = df_drop.dropna()

print("After dropping missing values:")
display(df_drop)
```

```python
#Fill missing values:
df_fill["Age"] = df_fill["Age"].fillna(df_fill["Age"].mean())

print("After filling missing Age values:")
display(df_fill)
```
