```python
import time
import numpy as np

# Define the size of the collections for performance comparison
size = 10**6 # One million elements

# --- Python List Performance ---
# Measure time taken for element-wise addition using Python lists
start_time_list = time.time()
list1 = list(range(size))
list2 = list(range(size))
result_list = [list1[i] + list2[i] for i in range(size)]
end_time_list = time.time()
time_list = end_time_list - start_time_list
print(f"Time taken by Python lists: {time_list:.6f} seconds")

# --- NumPy Array Performance ---
# Measure time taken for element-wise addition using NumPy arrays
start_time_numpy = time.time()
numpy_array1 = np.arange(size)
numpy_array2 = np.arange(size)
result_numpy = numpy_array1 + numpy_array2
end_time_numpy = time.time()
time_numpy = end_time_numpy - start_time_numpy
print(f"Time taken by NumPy arrays: {time_numpy:.6f} seconds")

# Compare performance
print(f"\nNumPy is approximately {time_list / time_numpy:.2f} times faster than Python lists for this operation.")
```

```python
import sys
import numpy as np

# Define the size of the collections for memory comparison
size = 10**6 # One million elements

# --- Python List Memory Usage ---
# Get the memory size of a Python list of integers
list_memory = sys.getsizeof(list(range(size)))
print(f"Memory taken by Python list (for {size} integers): {list_memory} bytes")

# --- NumPy Array Memory Usage ---
# Get the memory size of a NumPy array of integers
numpy_array_memory = np.arange(size).nbytes
print(f"Memory taken by NumPy array (for {size} integers): {numpy_array_memory} bytes")

# Compare memory usage
print(f"\nPython list uses approximately {list_memory / numpy_array_memory:.2f} times more memory than NumPy array for this data type and size.")
```

```python
# Create a 1-D array with numbers from 112 up to (but not including) 121
arr1=np.arange(112,121)
print(arr1)
```

```python
# Replace occurrences of the value 112 in 'arr1' with 111
arr1[arr1 == 112] = 111
print(arr1)
```

```python
# Attempt to replace numerical values with a string 'csd' in 'arr1' (this will cause a ValueError if the array is of integer type)
arr1[arr1==111]='csd'
print(arr1)
```

```python
# Corrected way to modify elements where arr1 is 112, 113, or 114
# Select elements that are equal to 112 OR 113 OR 114 using boolean indexing
condition = (arr1 == 112) |(arr1==113)|(arr1 == 114)

# Apply the multiplication only to the elements that satisfy the condition
arr1[condition] = arr1[condition] * 111

print(arr1)
```
